```python
M1 = np.random.randint(0, 51, size=(5, 10))    # 5 行 10 列
M2 = np.random.randint(0, 51, size=(10, 5))    # 10 行 5 列

print(M1)
print(M2)
```

2.2 Write a function to do matrix multiplication, i.e.Matrix_multip, M1 * M2. Here you are ONLY allowed to use for loop, operator *, and operator +.

```python
def Matrix_multip(M1, M2):
    rows = len(M1)         # 最终行数
    cols = len(M2[0])      # 最终列数
    n = len(M2)            # 中间值

    result = []
    for i in range(rows):
        row = []
        for j in range(cols):
            total = 0
            for k in range(n):
                total = total + int(M1[i][k]) * int(M2[k][j])
            row.append(total)
        result.append(row)

    return result
```

```python
print("M1 * M2 = ")
for row in Matrix_multip(M1, M2):
    print(row)
```

### 3. Pascal triangle

One of the most interesting number patterns is Pascal’s triangle (named after Blaise Pascal). Write a function with an argument to print the k Pascal_triangle kth line of the Pascal triangle. Report Pascal_triangle(100) and Pascal_triangle(200) .

```python
def Pascal_triangle(t):
    row = [1]                 # 第 0 行 [1]

    for i in range(1, t + 1):
        new_row = [1]         # 每行开头是 1
        for j in range(1, i):
            new_row.append(row[j - 1] + row[j])     # 中间的数 = 上一行相邻两个数之和
        new_row.append(1)     # 每行结尾是 1
        row = new_row

    return row
```

```python
print("Pascal_triangle(100)：")
print(Pascal_triangle(100))

print("\nPascal_triangle(200)：")
print(Pascal_triangle(200))
```

### 4. Add or double

If you start with 1 RMB and, with each move, you can either double your money or add another 1 RMB, what is the smallest number of moves you have to make to get to exactly x RMB? Here x is an integer randomly selected from 1 to 100. Write a function Least_moves to print your results. For example, Least_moves(2) should print 1, and Least_moves(5) should print 3.

```python
import random
```

```python
def Least_moves(t):
    n = 0
    while t != 1:
        if t % 2 != 0:
            t = t - 1
        else:
            t = t // 2
        n = n + 1

    return n
```

```python
x = random.randint(1, 100)

print("x为：" + str(x))
print("最小操作次数为：" + str(Least_moves(x)))
```

### 5. Dynamic programming

Insert + or - operation anywhere between the digits 1 2 3 4 5 6 7 8 9 in a way that the expression evaluates to an integer number. You may join digits together to form a bigger number. However, the digits must stay in the original order.

5.1 Write a function Find_expression, which should be able to print every possible solution that makes the expression evaluate to a random integer from 1 to 100. For example, Find_expression(50) should print lines including:  \[ 1 - 2 + 34 + 5 + 6 + 7 + 8 - 9 = 50 \] and \[ 1 + 2 + 34 - 56 + 78 - 9 = 50 \]

```python
import random
```

```python
def Find_expression(target):
    digits = "123456789"

    def make_expr(index, current):
        if index == len(digits):
            if eval(current) == target:
                print(current + " = " + str(target))
            return

        make_expr(index + 1, current + "+" + digits[index])   # 加
        make_expr(index + 1, current + "-" + digits[index])   # 减
        make_expr(index + 1, current + digits[index])         # 拼接

    make_expr(1, digits[0])
```

```python
target = random.randint(1, 100)
print(f"目标数：{target}")
Find_expression(target) 
```

5.2 Count the total number of suitable solutions for any integer i from 1 to 100, assign the count to a list called Total_solutions. Plot the list Total_solutions, so which number(s) yields the maximum and minimum of Total_solutions?

```python
import matplotlib.pyplot as plt
```

```python
def count_solutions(target):
    digits = "123456789"
    count = 0

    def make_expr(index, current):
        nonlocal count
        if index == len(digits):
            if eval(current) == target:
                count = count + 1
            return
        make_expr(index + 1, current + "+" + digits[index])
        make_expr(index + 1, current + "-" + digits[index])
        make_expr(index + 1, current + digits[index])

    make_expr(1, digits[0])
    return count
```

```python
Total_solutions = []
for i in range(1, 101):
    Total_solutions.append(count_solutions(i))
```

```python
plt.plot(range(1, 101), Total_solutions)
plt.xlabel("i")
plt.ylabel("count")
plt.title("Total_solutions")
plt.show()

print("最大:", max(Total_solutions))
print("最小:", min(Total_solutions))
```
